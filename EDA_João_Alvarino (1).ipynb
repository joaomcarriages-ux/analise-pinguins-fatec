# 3 Conheça as medidas numéricas

**Antes de pedir**

Agora vamos ler medidas-resumo. **Mínimo e máximo** delimitam os valores observados. **Média** é a soma dos valores dividida pela quantidade de valores válidos. **Mediana** é o valor central após ordenar os dados. O **primeiro quartil** marca a posição de 25% e o **terceiro quartil**, a de 75% dos dados ordenados.

Os quartis não são as perguntas Q1 e Q3. Para evitar confusão, peça os nomes por extenso. Uma medida-resumo não explica, sozinha, por que um valor ocorre.

**Pedido à IA**

Gere código para uma tabela com uma linha por variável e as colunas: quantidade de valores válidos, quantidade de ausentes, mínimo, primeiro quartil, mediana, terceiro quartil, máximo e média. Use nomes legíveis e arredonde as medidas para duas casas decimais. Não substitua ausências por zero e não modifique df. Explique como ler uma linha da tabela, sem escrever conclusões sobre os pinguins.


```python
# Seleção das colunas numéricas de interesse
colunas_numericas = ['bill_length_mm', 'bill_depth_mm', 'flipper_length_mm', 'body_mass_g']

# Criando um dicionário para armazenar as estatísticas calculadas individualmente
estatisticas = {}

for col in colunas_numericas:
    estatisticas[col] = {
        'Válidos': df[col].notnull().sum(),
        'Ausentes': df[col].isnull().sum(),
        'Mínimo': df[col].min(),
        'Primeiro Quartil': df[col].quantile(0.25),
        'Mediana': df[col].median(),
        'Terceiro Quartil': df[col].quantile(0.75),
        'Máximo': df[col].max(),
        'Média': df[col].mean()
    }

# Convertendo o resultado em um DataFrame e arredondando para duas casas decimais
tabela_resumo = pd.DataFrame(estatisticas).T
tabela_resumo = tabela_resumo.round(2)

# Exibindo a tabela resumo de forma organizada
display(tabela_resumo)
```

**Pedido à IA**

> Quero comparar a distribuição de massa corporal entre espécies. Minha tabela é df. A coluna com a espécie é [preencha] e a coluna com a massa em gramas é [preencha]. Gere código para criar dados_q2 com valores disponíveis nessas duas colunas, preservando df. Informe o número de registros utilizados e não utilizados. Mostre uma tabela por espécie com quantidade válida, mínimo, primeiro quartil, mediana, terceiro quartil, máximo e média de massa. Arredonde para duas casas decimais. Ainda não faça o gráfico nem responda à questão por mim.


```python
import pandas as pd
import os

# Garantir que df esteja carregado na sessão atual
if 'df' not in locals() and 'df' not in globals():
    caminho_arquivo = '/content/drive/MyDrive/Colab Notebooks/datasets/penguins.csv'
    if os.path.exists(caminho_arquivo):
        df = pd.read_csv(caminho_arquivo)
    else:
        # Tentativa alternativa local caso o drive não esteja montado no mesmo caminho
        df = pd.read_csv('https://raw.githubusercontent.com/mwaskom/seaborn-data/master/penguins.csv')

# 1. Filtrar registros válidos que possuem tanto 'species' quanto 'body_mass_g'
dados_q2 = df[df['species'].notnull() & df['body_mass_g'].notnull()].copy()

# 2. Informar o número de registros utilizados e não utilizados
total_original = len(df)
total_utilizado = len(dados_q2)
fora_analise = total_original - total_utilizado

print(f"Registros utilizados na análise (Q2): {total_utilizado}")
print(f"Registros fora da análise (sem espécie ou sem massa): {fora_analise}\n")

# 3. Calcular as estatísticas descritivas agrupadas por espécie
resumo_q2 = dados_q2.groupby('species')['body_mass_g'].agg(
    Quantidade_Valida='count',
    Minimo='min',
    Primeiro_Quartil=lambda x: x.quantile(0.25),
    Mediana='median',
    Terceiro_Quartil=lambda x: x.quantile(0.75),
    Maximo='max',
    Media='mean'
)

# Renomear as colunas para melhor legibilidade
resumo_q2.columns = [
    'Quantidade válida',
    'Mínimo',
    'Primeiro Quartil',
    'Mediana',
    'Terceiro Quartil',
    'Máximo',
    'Média'
]

# Arredondar para duas casas decimais e exibir a tabela
resumo_q2 = resumo_q2.round(2)
display(resumo_q2)
```

## Antes do próximo pedido — Como ler um boxplot

Um **boxplot** resume a distribuição de uma medida. No próximo gráfico, haverá uma caixa por espécie, com a massa em gramas no eixo vertical.

- A linha dentro da caixa indica a **mediana**.
- A parte inferior da caixa indica o **primeiro quartil** e a superior, o **terceiro quartil**. A caixa representa a faixa central de aproximadamente metade dos valores.
- As linhas que se estendem da caixa, chamadas hastes, seguem um critério de distância baseado nos quartis. Elas não são necessariamente o mínimo e o máximo.
- Pontos além das hastes são valores sinalizados por esse critério. Não são automaticamente erros e não devem ser excluídos só por aparecerem assim.
